```python
import numpy as np
import pylab as pl
import matplotlib as mpl
import matplotlib.patches as patches
```

```python
# Eq. 2: The holographic mask

def H(psiref, psiobject):
  ans = np.abs(psiref + psiobject)**2
  return ans

def Intensity(psi_illumination, mask):
  return np.abs(np.fft.fftshift(np.fft.fft2(psi_illumination*mask)))
```

```python
x = np.linspace(-1,1,500)*5
y = np.linspace(-1,1,500)[::-1]*5
X,Y = np.meshgrid(x,y)
lam = 0.632 # nm
k = 2*np.pi/lam

psiref = lambda k: np.exp(1j*k*X)
psiobject = lambda m: np.exp(1j*m*np.arctan2(Y,X))

fig = pl.figure(figsize=(15,10))
ax = []
for m in range(1,7):
  H1 = H(psiref(k), psiobject(m))
  ax.append(fig.add_subplot(2,3,m))
  ax[m-1].imshow(H1, cmap='gray')
  ax[m-1].axis('off')
  ax[m-1].text(30, 24, [f' {i} ' for i in list('abcdef')][m-1], fontsize=24, color="#ffffff", horizontalalignment='center', verticalalignment='center', bbox=dict(facecolor='#000000', alpha=1, boxstyle='round,pad=0'))

```

Fig. 2 a-f

```python
w = 0.8
psiG = np.exp(-(X*X+Y*Y)/(w*w))

fig1 = pl.figure(figsize=(15,5))
ax1 = fig1.add_subplot(1,1,1)

kx = k
m = 1
H1 = H(psiref(k), psiobject(m))
# ax1.imshow(psiG, cmap='gray')
ax1.contourf(x,y, Intensity(psiG, H1), cmap='gray', levels=100)
ax1.axis('off')
ax1.set_xlim([-0.5, 0.5])
ax1.set_ylim([-0.25,0.25])
ax1.set_aspect(1)
ax1.text(-0.5, 0.2, "g", fontsize=20, color="#ffffff")
ax1.text(-0.3, -0.2, r"$m=-1$", fontsize=20, color="#ffffff", horizontalalignment='center', verticalalignment='center')
ax1.text(0.35, -0.2, r"$m=1$", fontsize=20, color="#ffffff", horizontalalignment='center', verticalalignment='center')
ax1.text(0.0, -0.2, r"$m=0$", fontsize=20, color="#ffffff", horizontalalignment='center', verticalalignment='center')

style = "Simple, tail_width=0.5, head_width=4, head_length=8"
kw = dict(arrowstyle=style, color="white")
aright = patches.FancyArrowPatch( (-0.4, 0.1), (-0.2, 0.1),
                             connectionstyle="arc3,rad=-.5", **kw)
aleft = patches.FancyArrowPatch((0.42, 0.1), (0.22, 0.1),
                             connectionstyle="arc3,rad=.5", **kw)
pl.gca().add_patch(aright)
pl.gca().add_patch(aleft)

```
